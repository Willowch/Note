```python
import matplotlib.pyplot as plt
from matplotlib.axes import Axes


plt.rcParams['font.sans-serif']=['SimHei']#设置中文以黑体输出

#创建子图:figure为画布对象;axes为子图数组,Axes类型
figure,axes=plt.subplots(nrows=1,ncols=2,figsize=(20,6),dpi=100)

countries=['中国','美国','英国','法国','德国','日本','印度']
values=[45,30,25,20,15,10,5]

axes1: Axes=axes[0]#重命名,加上注解以便提示信息
axes1.bar(countries,values,width=0.6,color='g')#创建柱状图,设置宽度和颜色
axes1.set_title('世界石油储备',fontsize=16)
axes1.set_xlabel('国家',fontsize=13)
axes1.set_ylabel('石油储备(亿吨)',fontsize=13)

axes1.grid(linestyle='--',alpha=0.3)




countries2=['中国','美国','英国','法国','德国','日本','印度']
values2=[45,30,25,20,15,10,5]

axes2: Axes=axes[1]
axes2.pie(values2,labels=countries2,autopct='%1.1f%%')#中间符号为显示,两侧为格式;前1为最小长度(不够前面添空格),后1为保留小数位;
axes2.set_title('世界石油储备占比',fontsize=16)
axes2.legend(loc='lower center',ncol=5,bbox_to_anchor=(0.5,-0.08))#ncol设置图标显示列数;
#bbox:0.5表示左右中间,-0.08表示上下方向-0.08的位置;


plt.savefig('data/plt_test.png')
plt.show()




```
